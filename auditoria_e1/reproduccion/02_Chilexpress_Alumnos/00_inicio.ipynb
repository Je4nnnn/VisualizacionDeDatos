# Chilexpress cumplimiento y capacidad de reparto
Datos simulados docentes. Abrir desde la carpeta del caso. Este inicio verifica archivos y carga textos sin resolver la limpieza.

```python
from pathlib import Path
import json, hashlib
import pandas as pd
BASE = Path.cwd()
meta = json.loads((BASE / "metadatos.json").read_text(encoding="utf-8"))
for f in meta["files"]:
    assert hashlib.sha256((BASE / f["archivo"]).read_bytes()).hexdigest() == f["sha256"]
lotes = {}
for lote in ["base", "actualizacion", "ampliacion"]:
    ruta = BASE / "datos" / lote / (meta["main"] + ".csv")
    lotes[lote] = pd.read_csv(ruta, dtype="string")
    print(lote, lotes[lote].shape)
# No concatenar tablas complementarias como si fueran observaciones principales.
```
